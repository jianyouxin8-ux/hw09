## 作業簡述
針對「桌遊體驗」與「電影欣賞」兩場活動的逗號分隔報名資料，先清理半形空白與大小寫並拆成串列，再搭配 `tuple`、`list`、`set`、`dict` 處理去重、交集與差集比對、排序切片及費用計算，最後印出完整摘要。

## 測試一：原始資料

```python
# 固定活動名稱（tuple）與收費字典（dict）
activity_names = ("桌遊體驗", "電影欣賞")
game_label = activity_names[0]
movie_label = activity_names[1]
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

# 原始報名文字（保留原值）
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "

# 去半形空白、轉小寫、依逗號切分，保留原順序與重複項目
game_entries = board_raw.replace(" ", "").lower().split(",")
movie_entries = movie_raw.replace(" ", "").lower().split(",")

# 轉為集合去重並計算交集、聯集與雙向差集
game_set = set(game_entries)
movie_set = set(movie_entries)
both_participants = game_set & movie_set
union_participants = game_set | movie_set
game_only = game_set - movie_set
movie_only = movie_set - game_set

# 排序後用切片取通知名單前兩位
all_sorted = sorted(union_participants)
top_two_notice = all_sorted[:2]

# 彙整人數、人次與應收金額至 summary 字典
summary = {
    "game_num": len(game_set),
    "movie_num": len(movie_set),
    "both_num": len(both_participants),
    "unique_total": len(union_participants),
    "total_visits": len(game_set) + len(movie_set),
    "game_amount": len(game_set) * unit_fees[game_label],
    "movie_amount": len(movie_set) * unit_fees[movie_label],
}
summary["grand_total"] = summary["game_amount"] + summary["movie_amount"]

# 輸出摘要
print("HW09 活動報名摘要")
print(f"活動：{game_label}，每人 {unit_fees[game_label]:.2f} 元")
print(f"活動：{movie_label}，每人 {unit_fees[movie_label]:.2f} 元")
print(f"桌遊整理串列：{game_entries}")
print(f"電影整理串列：{movie_entries}")
print(f"桌遊去重名單：{sorted(game_set)}")
print(f"電影去重名單：{sorted(movie_set)}")
print(f"桌遊人數：{summary['game_num']} 人")
print(f"電影人數：{summary['movie_num']} 人")
print(f"共同參加者：{sorted(both_participants)}")
print(f"共同人數：{summary['both_num']} 人")
print(f"只參加桌遊者：{sorted(game_only)}")
print(f"只參加電影者：{sorted(movie_only)}")
print(f"全部參加者：{all_sorted}")
print(f"總不重複人數：{summary['unique_total']} 人")
print(f"報名人次：{summary['total_visits']} 人次")
print(f"通知名單前兩位：{top_two_notice}")
print(f"桌遊應收：{summary['game_amount']:.2f} 元")
print(f"電影應收：{summary['movie_amount']:.2f} 元")
print(f"總應收：{summary['grand_total']:.2f} 元")
```

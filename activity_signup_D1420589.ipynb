```python
# HW09｜兩場活動報名整理器
# 姓名：黃星瑜｜學號：D1420589

# 讀取固定活動名稱（tuple）與收費標準（dict）
activity_names = ("桌遊體驗", "電影欣賞")
game_title = activity_names[0]
film_title = activity_names[1]

unit_fees = {
    "桌遊體驗": 120,
    "電影欣賞": 80,
}

# 保留原始報名文字（str）
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "  # 測試二：" ben, Dora, ERIC, ben, Frank " ｜ 測試三：" AMY, ben, CORA, amy "

# 清理半形空白與大小寫後拆成串列，保留原始順序與重複報名
clean_game_list = board_raw.replace(" ", "").lower().split(",")
clean_film_list = movie_raw.replace(" ", "").lower().split(",")

# 轉成集合去重，並以集合方法比對兩場名單
game_unique = set(clean_game_list)
film_unique = set(clean_film_list)

shared_members = game_unique.intersection(film_unique)      # 兩場都參加
all_members = game_unique.union(film_unique)                # 至少參加一場
game_only_members = game_unique.difference(film_unique)     # 只參加桌遊
film_only_members = film_unique.difference(game_unique)     # 只參加電影

# 依字母順序排序，並切出通知名單前兩位
all_members_sorted = sorted(all_members)
notify_top2 = all_members_sorted[:2]

# 將統計與收費存入 summary 字典（先算人數，再從字典取值算人次與金額）
summary = {
    "game_count": len(game_unique),
    "film_count": len(film_unique),
    "shared_count": len(shared_members),
    "unique_total": len(all_members),
}
summary["signup_person_times"] = summary["game_count"] + summary["film_count"]
summary["game_fee"] = summary["game_count"] * unit_fees[game_title]
summary["film_fee"] = summary["film_count"] * unit_fees[film_title]
summary["total_fee"] = summary["game_fee"] + summary["film_fee"]

# 印出完整摘要
print("HW09 活動報名摘要")
print(f"活動：{game_title}，每人 {unit_fees[game_title]:.2f} 元")
print(f"活動：{film_title}，每人 {unit_fees[film_title]:.2f} 元")
print(f"桌遊整理串列：{clean_game_list}")
print(f"電影整理串列：{clean_film_list}")
print(f"桌遊去重名單：{sorted(game_unique)}")
print(f"電影去重名單：{sorted(film_unique)}")
print(f"桌遊人數：{summary['game_count']} 人")
print(f"電影人數：{summary['film_count']} 人")
print(f"共同參加者：{sorted(shared_members)}")
print(f"共同人數：{summary['shared_count']} 人")
print(f"只參加桌遊者：{sorted(game_only_members)}")
print(f"只參加電影者：{sorted(film_only_members)}")
print(f"全部參加者：{all_members_sorted}")
print(f"總不重複人數：{summary['unique_total']} 人")
print(f"報名人次：{summary['signup_person_times']} 人次")
print(f"通知名單前兩位：{notify_top2}")
print(f"桌遊應收：{summary['game_fee']:.2f} 元")
print(f"電影應收：{summary['film_fee']:.2f} 元")
print(f"總應收：{summary['total_fee']:.2f} 元")
```
